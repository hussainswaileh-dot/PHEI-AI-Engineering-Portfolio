```python
import sys
import pandas as pd

print("PHEI Experiment 001")
print("Python:", sys.version.split()[0])
print("Pandas:", pd.__version__)
print("Environment: READY")
```

```python

data = {
    "Manifest_ID": ["MW001", "MW002", "MW003"],
    "Facility": ["Hospital_A", "Hospital_B", "Hospital_C"],
    "Transport_Weight": [120, 85, 150],
    "Treatment_Weight": [120, 80, 150]
}

df = pd.DataFrame(data)

display(df)
```

```python

df["Weight_Difference"] = (
    df["Transport_Weight"] -
    df["Treatment_Weight"]
)

display(df)
```

```python

exceptions = df[df["Weight_Difference"] != 0]

print("Total Records:", len(df))
print("Exceptions Found:", len(exceptions))

display(exceptions)
```

```python

assert len(df) == 3
assert len(exceptions) == 1

assert exceptions.iloc[0]["Manifest_ID"] == "MW002"
assert exceptions.iloc[0]["Weight_Difference"] == 5

print("ALL 4 TESTS PASSED")
print("Experiment 001: VALIDATED")
```
